## 3. Example Pi2 Usage: File I/O and Image Binning

With Pi2 initialized, let's explore some basic image processing operations. The following examples demonstrate reading an image, performing binning, and saving results.

> **Prerequisite:** For these examples to run, you'll need an image file named `t1-head.tif` located in your `/content/drive/MyDrive/pi2/` directory. If you don't have this specific file, you can upload any TIFF image to that location and adjust the `filename` in the `input_file()` function below.

#### 3.1 Helper Functions for File Paths

These Python functions are defined to make it easier to specify input and output file paths within your `MyDrive/pi2/` directory.

```python
def input_file(filename='t1-head.tif'):
        return '/content/drive/MyDrive/pi2/' + filename

def input_file_bin():
        return '/content/drive/MyDrive/pi2/t1-head_bin_256x256x129.raw'

def output_file(name):
        return '/content/drive/MyDrive/pi2/' + name
```

#### 3.2 Example Function: Binning Operations

This function, `binning_scaling()`, demonstrates how to:
1.  Read an image using `pi.read()`.
2.  Create new images with the same properties as the input using `pi.newlike()`.
3.  Perform image binning with different modes (mean, min, max) using `pi.bin()`.
4.  Save the processed images using `pi.writeraw()`. (Note: `writeraw` saves without a specific format header, useful for simple array data. `writetif` is generally preferred for TIFFs.)

```python
def binning_scaling():
        """
        Demonstrates binning.
        """

        # Read image
        img = pi.read(input_file())

        # Normal binning
        binned_mean = pi.newlike(img)
        pi.bin(img, binned_mean, 4)
        pi.writeraw(binned_mean, output_file('binning_mean'))

        # Min binning
        binned_min = pi.newlike(img)
        pi.bin(img, binned_min, 4, 'min')
        pi.writeraw(binned_min, output_file('binning_min'))

        # Max binning
        binned_max = pi.newlike(img)
        pi.bin(img, binned_max, 4, 'max')
        pi.writeraw(binned_max, output_file('binning_max'))
```

#### 3.3 Run the Binning Example

Now, let's execute the `binning_scaling()` function. This will process the `t1-head.tif` image (or your specified input) and save three binned versions (`binning_mean.raw`, `binning_min.raw`, `binning_max.raw`) to your `/content/drive/MyDrive/pi2/` directory.

```python
binning_scaling()
```
